```python
from pathlib import Path
import re

root_dir = Path(r"H:\Projects\speech_synthesis\speech_synthesis\samples\ALS\SpeedPerturbation\SP")

for patient_dir in root_dir.iterdir():
    if not patient_dir.is_dir():
        continue

    for wav_file in patient_dir.glob("*.wav"):

        match = re.match(r"^(\d+_Patient_\d+)", wav_file.stem)

        if match:
            new_name = match.group(1) + ".wav"
            new_path = wav_file.parent / new_name

            # Avoid accidentally overwriting an existing file
            if new_path.exists():
                print(f"SKIPPED: {new_name} already exists")
                continue

            wav_file.rename(new_path)
            print(f"Renamed: {wav_file.name} -> {new_name}")
```
