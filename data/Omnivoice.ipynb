```python
from omnivoice import OmniVoice
import soundfile as sf
import torch

model = OmniVoice.from_pretrained(
    "k2-fsa/OmniVoice",
    device_map="cuda:0",
    dtype=torch.float16
)

ref_audio = r"H:\Projects\speech_synthesis\speech_synthesis\samples\ALS\Real\Patient_48\001_Patient_48_10302014190442_recording.wav"

sentence = "I owe you a yoyo today."

audio = model.generate(
    text=sentence,
    ref_audio=ref_audio,
    ref_text=sentence,
)

sf.write(
    r"H:\Projects\speech_synthesis\speech_synthesis\samples\ALS\OmniVoice\Patient_48\001_Patient_48_10302014190442_recording.wav",
    audio[0],
    24000
)
```
