```python
import librosa

speed_factor = 0.8

# Effective resampling rate
target_sr = round(sr / speed_factor)

perturbed_audio = librosa.resample(
    audio,
    orig_sr=sr,
    target_sr=target_sr,
    axis=0
)

print("Original duration :", len(audio) / sr)
print("Perturbed duration:", len(perturbed_audio) / sr)
```

```python
sf.write(
    output_audio,
    perturbed_audio,
    sr
)

print("Saved to:")
print(output_audio)
```

```python
from pathlib import Path
import soundfile as sf
import librosa

# ============================================================
# SETTINGS
# ============================================================

input_root = Path(
    r"H:\Projects\speech_synthesis\speech_synthesis\samples\ALS\Real"
)

output_root = Path(
    r"H:\Projects\speech_synthesis\speech_synthesis\samples\ALS\SpeedPerturbation"
)

speed_factor = 0.9

# Separate folder for this perturbation factor
factor_folder = output_root / str(speed_factor)
factor_folder.mkdir(parents=True, exist_ok=True)


# ============================================================
# FIND PATIENT FOLDERS
# ============================================================

patient_folders = sorted(
    folder
    for folder in input_root.iterdir()
    if folder.is_dir()
)

print(f"Found {len(patient_folders)} patient folders.")
print(f"Speed factor: {speed_factor}")
print(f"Output: {factor_folder}\n")


# ============================================================
# PROCESS ALL PATIENTS
# ============================================================

total_processed = 0
failed_files = []

for patient_index, patient_folder in enumerate(
    patient_folders,
    start=1
):

    wav_files = sorted(patient_folder.glob("*.wav"))

    print(
        f"[{patient_index}/{len(patient_folders)}] "
        f"{patient_folder.name} "
        f"({len(wav_files)} recordings)"
    )

    # Create corresponding patient folder
    output_patient_folder = (
        factor_folder / patient_folder.name
    )

    output_patient_folder.mkdir(
        parents=True,
        exist_ok=True
    )

    # --------------------------------------------------------
    # Process each recording
    # --------------------------------------------------------

    for input_audio in wav_files:

        try:
            # Load audio
            audio, sr = sf.read(input_audio)

            # Calculate effective resampling rate
            target_sr = round(
                sr / speed_factor
            )

            # Speed perturbation
            perturbed_audio = librosa.resample(
                audio,
                orig_sr=sr,
                target_sr=target_sr,
                axis=0
            )

            # Preserve original filename
            output_audio = (
                output_patient_folder
                / input_audio.name
            )

            # Save using original sampling rate
            sf.write(
                output_audio,
                perturbed_audio,
                sr
            )

            total_processed += 1

        except Exception as error:

            failed_files.append(
                (input_audio, str(error))
            )

            print(
                f"    ERROR: {input_audio.name}"
            )


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("FINISHED")
print("=" * 60)

print(f"Patients: {len(patient_folders)}")
print(f"Successfully processed: {total_processed}")
print(f"Failed: {len(failed_files)}")

if failed_files:

    print("\nFailed files:")

    for file_path, error in failed_files:
        print(f"\n{file_path}")
        print(f"    {error}")
```
